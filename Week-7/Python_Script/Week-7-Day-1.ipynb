## **CLUSTERING**
Clustering is a fundamental task in unsupervised machine learning that involves grouping data points such that those within the same group are more similar to each other than to those in other groups. The goal is to uncover hidden structures in data without relying on labeled outcomes. Among the most widely used clustering techniques are **K-Means**, **Hierarchical Clustering**, and **DBSCAN**. Each of these algorithms follows a distinct philosophy in how it defines and discovers clusters, making them suitable for different types of data and analytical objectives.

## **K-Means Clustering**

K-Means is one of the simplest and most commonly used clustering algorithms. It operates under the assumption that;

* Clusters are spherical
* well-separated
* roughly equal size.

The algorithm seeks to partition a dataset into *k* distinct, non-overlapping clusters by minimizing the within-cluster sum of squares (WCSS), which is the sum of squared distances between each point and the centroid of its assigned cluster.

The process begins by selecting the number of clusters, *k*, which must be specified in advance. Initial centroids are chosen either randomly or using a more strategic method such as K-Means++. Each data point is then assigned to the nearest centroid based on Euclidean distance. After all points are assigned, the centroids are updated to the mean of the points in each cluster. This assignment and update cycle repeats iteratively until the centroids stabilize—meaning that further iterations do not result in significant changes in cluster membership.

One of the strengths of K-Means lies in its computational efficiency and scalability, making it suitable for large datasets. It is often used in applications such as customer segmentation, image compression, and document clustering. However, its performance depends heavily on the choice of *k*. To determine the optimal number of clusters, practitioners commonly use the **elbow method**, which involves plotting WCSS against different values of *k* and identifying the point where the rate of decrease sharply slows—a visual "elbow." Alternatively, the **silhouette score**, which measures how tightly grouped data points are within clusters compared to other clusters, can provide a more quantitative assessment.

Despite its popularity, K-Means has notable limitations. It is sensitive to the initial placement of centroids, which can lead to convergence on suboptimal solutions. Multiple random restarts (controlled by the `n_init` parameter in implementations) help mitigate this issue. Additionally, K-Means struggles with clusters that are not convex or spherical in shape, and it is easily influenced by outliers due to its reliance on mean-based centroids. It also assumes that all clusters are of similar size, which may not reflect real-world data distributions.

## **Hierarchical Clustering**

Hierarchical clustering offers a different approach by building a hierarchy of clusters, either through an agglomerative (bottom-up) or divisive (top-down) strategy. The most common variant is **agglomerative clustering**, in which each data point starts as its own cluster, and pairs of clusters are progressively merged based on a chosen linkage criterion until all points belong to a single cluster.

The merging process is guided by a distance metric—typically Euclidean distance—and a **linkage rule** that defines how the distance between clusters is computed. Single linkage uses the shortest distance between any two points in different clusters, which can lead to elongated, chain-like clusters. Complete linkage uses the maximum distance, favoring compact clusters. Average linkage computes the average distance between all pairs of points across clusters, offering a balance between the two. **Ward’s method**, one of the most effective linkage criteria, minimizes the total within-cluster variance and tends to produce compact, spherical clusters similar to K-Means.

The result of hierarchical clustering is naturally represented as a **dendrogram**, a tree-like diagram that illustrates the sequence of merges and the distances at which they occur. This visualization allows analysts to inspect the clustering structure and choose an appropriate number of clusters by cutting the tree at a desired height. Unlike K-Means, hierarchical clustering does not require the number of clusters to be specified in advance, offering greater flexibility in exploratory analysis.

However, this method comes with computational costs. The time complexity is generally O(n³), making it less suitable for very large datasets. Additionally, once a merge is made, it cannot be undone, which means the algorithm is deterministic and lacks the ability to correct earlier decisions. Despite these drawbacks, hierarchical clustering is highly interpretable and widely used in fields such as biology (e.g., phylogenetic trees), social network analysis, and taxonomy construction where understanding the relationships between clusters is as important as the clusters themselves.

## **DBSCAN (Density-Based Spatial Clustering of Applications with Noise)**

DBSCAN takes a fundamentally different approach by defining clusters based on **density** rather than distance to centroids or hierarchical merging. It identifies clusters as dense regions of data points separated by areas of lower density. This allows DBSCAN to discover clusters of arbitrary shape and to distinguish between core points, border points, and noise.

The algorithm is governed by two key parameters: **ε (epsilon)**, which defines the radius around a point to search for neighbors, and **MinPts**, the minimum number of points required within that radius for a point to be considered a core point. A core point is one that has at least MinPts neighbors within distance ε. All points within ε of a core point belong to the same cluster, even if they are not core points themselves—these are called border points. Points that are neither core nor border are labeled as **noise** or outliers.

One of the major advantages of DBSCAN is its ability to detect outliers, making it particularly useful in anomaly detection tasks. It also does not require the number of clusters to be specified in advance, and it can naturally identify clusters of varying shapes and sizes. This makes it robust in real-world scenarios where data may not conform to idealized geometric assumptions.

However, DBSCAN is sensitive to the choice of ε and MinPts. Selecting inappropriate values can lead to either over-merging of clusters or excessive fragmentation. A common strategy for choosing ε is to use a **k-distance graph**, where the k-th nearest neighbor distance is plotted for each point; the "elbow" in this plot often suggests a suitable ε value. DBSCAN also struggles when clusters have widely varying densities, as a single ε may not work well across the entire dataset. Additionally, performance degrades in high-dimensional spaces due to the "curse of dimensionality," where distances between points become less meaningful.

Despite these challenges, DBSCAN is widely used in spatial data analysis, such as identifying geographical hotspots, detecting fraudulent transactions, and clustering GPS trajectories. Its robustness to noise and flexibility in cluster shape make it a powerful tool when the underlying data structure is complex or poorly understood.

## **Conclusion**

K-Means, Hierarchical Clustering, and DBSCAN each offer unique strengths and are suited to different types of data and analytical goals. K-Means is efficient and effective for well-behaved, spherical clusters but fails when assumptions are violated. Hierarchical clustering provides interpretability and a natural hierarchy but is computationally expensive. DBSCAN excels at finding arbitrarily shaped clusters and identifying noise but requires careful parameter tuning. Understanding the principles, assumptions, and limitations of each algorithm enables data scientists to choose the most appropriate method for their specific problem, ensuring meaningful and actionable insights from unlabeled data.

# 1️⃣ Creating the K-Means Model

```python
kmeans = KMeans(n_clusters=3, init='k-means++', n_init=25, random_state=123)
```

This line **creates a K-Means clustering model** with specific settings.

### Parameters explained

**1. `n_clusters=3`**

* The algorithm will divide the data into **3 clusters (groups)**.
* Each observation will belong to one of these clusters.

**2. `init='k-means++'`**

* Determines how the **initial cluster centers (centroids)** are selected.
* **k-means++** is a smart initialization method that spreads centroids apart.
* It usually leads to **faster convergence and better clustering results**.

**3. `n_init=25`**

* The algorithm will run **25 different times** with different centroid initializations.
* It then **selects the best clustering solution** based on the lowest **within-cluster sum of squares (inertia)**.
* More runs improve reliability.

**4. `random_state=42`**

* Fixes the **random seed** so that results are **reproducible**.
* If someone else runs the same code, they will get the **same clustering result**.

# What if you don’t know the number of clusters?

If you **have no idea how many clusters exist**, you use **cluster evaluation methods** to estimate a good value of (k).

The most common methods are:

---

# 1️⃣ Elbow Method

This is the **most widely used approach**.

The idea is to compute the **within-cluster sum of squares (WCSS)** for several values of (k).

Steps:

1. Run K-Means for (k = 1,2,3,4,...,10)
2. Calculate **inertia (WCSS)** for each (k)
3. Plot the results
4. Choose the **“elbow point”** where improvement slows.

Example code:

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []

for k in range(1,11):
    kmeans = KMeans(n_clusters=k, random_state=123)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,11), wcss, marker='o')
plt.xlabel("Number of clusters (k)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

Interpretation:

* The **bend (elbow)** of the curve indicates the **optimal number of clusters**.

---

# 2️⃣ Silhouette Method

Another very reliable technique is the **Silhouette Score**.

It measures:

* **How similar a point is to its own cluster**
* **How different it is from other clusters**

Score range:

* **+1 → well clustered**
* **0 → overlapping clusters**
* **−1 → wrong clustering**

Example:

```python
from sklearn.metrics import silhouette_score

for k in range(2,10):
    kmeans = KMeans(n_clusters=k, random_state=123)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    print(k, score)
```

Choose the **k with the highest score**.

---

Another method is the **Gap Statistic**, which compares clustering results with random data.

It is **more statistically rigorous** but less commonly used in beginner analysis.

### **📈 Elbow Method to Choose Optimal k**

````python
# Elbow method
inertias = []
K_range = range(1, 11)

for k in K_range:
    kmeans_temp = KMeans(n_clusters=k, init='k-means++', n_init=25, random_state=123)
    kmeans_temp.fit(X_scaled)
    inertias.append(kmeans_temp.inertia_)
# Plot
plt.figure(figsize=(8, 5))
plt.plot(K_range, inertias, 'bo-', linewidth=2, markersize=6)
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Within-Cluster Sum of Squares (WCSS)')
plt.grid(True)
plt.show()
````

### **Data Preprocessing**

````python
# Create a copy for preprocessing
df_clean = df.copy()
df_clean.head()
````

````python
# Drop rows with missing values (or you can impute them)
df_clean = df_clean.dropna()
print(f"\nRows after removing missing values: {len(df_clean)}")
````

````python
# Select numerical features for clustering
numerical_features = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']
numerical_features
````

````python
# Create feature matrix
X = df_clean[numerical_features].values
X
````

````python
# Store species for later comparison
y_true = df_clean['species'].values
y_true[:5]
````

````python
# Encode species to numerical values for comparison
le = LabelEncoder()
y_encoded = le.fit_transform(y_true)
y_encoded
````

````python
print(f"\nFeature matrix shape: {X.shape}")
print(f"\nEncoded species: {le.classes_}")
````

### **Feature Scaling**

````python
# Standardize features (important for K-Means)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("\n" + "=" * 60)
print("FEATURE SCALING COMPLETE")
print("=" * 60)
print(f"Mean after scaling: {X_scaled.mean(axis=0).round(4)}")
print(f"Std after scaling: {X_scaled.std(axis=0).round(4)}")
````

### **Determining the Optimal Number of K**

````python
# Elbow Method
inertias = []
K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertias.append(kmeans.inertia_)

# Silhouette Score
silhouette_scores = []
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    silhouette_scores.append(silhouette_score(X_scaled, labels))

# Davies-Bouldin Score
db_scores = []
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    db_scores.append(davies_bouldin_score(X_scaled, labels))
````

````python
# Plot the results
fig, axes = plt.subplots(1, 3, figsize=(18, 9))

# Elbow Plot
axes[0].plot(K_range, inertias, 'bo-', linewidth=2, markersize=8)
axes[0].set_xlabel('Number of Clusters (k)', fontsize=12)
axes[0].set_ylabel('Inertia', fontsize=12)
axes[0].set_title('Elbow Method', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)
axes[0].annotate('Elbow Point', xy=(3, inertias[1]), xytext=(5, inertias[1]+500),
                arrowprops=dict(arrowstyle='->', color='red'), fontsize=10)

# Silhouette Score Plot
axes[1].plot(K_range, silhouette_scores, 'go-', linewidth=2, markersize=8)
axes[1].set_xlabel('Number of Clusters (k)', fontsize=12)
axes[1].set_ylabel('Silhouette Score', fontsize=12)
axes[1].set_title('Silhouette Score', fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3)
axes[1].axhline(y=max(silhouette_scores), color='r', linestyle='--', alpha=0.5)

# Davies-Bouldin Score Plot
axes[2].plot(K_range, db_scores, 'ro-', linewidth=2, markersize=8)
axes[2].set_xlabel('Number of Clusters (k)', fontsize=12)
axes[2].set_ylabel('Davies-Bouldin Score', fontsize=12)
axes[2].set_title('Davies-Bouldin Score', fontsize=14, fontweight='bold')
axes[2].grid(True, alpha=0.3)
axes[2].axhline(y=min(db_scores), color='g', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.savefig('cluster_evaluation.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n" + "=" * 60)
print("OPTIMAL CLUSTER RECOMMENDATIONS")
print("=" * 60)
print(f"Best k by Silhouette Score: {K_range[silhouette_scores.index(max(silhouette_scores))]}")
print(f"Best k by Davies-Bouldin Score: {K_range[db_scores.index(min(db_scores))]}")
````

#### **Apply K-Means Clustering**

````python
# Based on evaluation, use k=3 (matches the 3 penguin species)
k = 3
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10, max_iter=300)
kmeans.fit(X_scaled)

# Get cluster labels
cluster_labels = kmeans.labels_

# Add cluster labels to dataframe
df_clean['cluster'] = cluster_labels

# Add cluster centers (inverse transform to original scale)
cluster_centers_scaled = kmeans.cluster_centers_
cluster_centers = scaler.inverse_transform(cluster_centers_scaled)

print("\n" + "=" * 60)
print("K-MEANS CLUSTERING RESULTS")
print("=" * 60)
print(f"\nNumber of clusters: {k}")
print(f"\nCluster centers (original scale):\n{pd.DataFrame(cluster_centers, columns=numerical_features)}")
print(f"\nCluster distribution:\n{df_clean['cluster'].value_counts().sort_index()}")
````

````python
# PCA for 2D visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create figure with two plots
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# 1. PCA Plot with Cluster Labels
axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=cluster_labels, cmap='viridis',
                s=100, alpha=0.6, edgecolors='black', linewidth=0.5)
axes[0].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%})', fontsize=12)
axes[0].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%})', fontsize=12)
axes[0].set_title('K-Means Clusters (PCA)', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)

# 2. PCA Plot with True Species
axes[1].scatter(X_pca[:, 0], X_pca[:, 1], c=y_encoded, cmap='Set2',
                s=100, alpha=0.6, edgecolors='black', linewidth=0.5)
axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%})', fontsize=12)
axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%})', fontsize=12)
axes[1].set_title('True Species (PCA)', fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

### **Best Practice**

````python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score,
    adjusted_rand_score,
    normalized_mutual_info_score,
    homogeneity_score
)

import pandas as pd


def perform_kmeans_clustering(data_path, n_clusters=3, random_state=42):

    # Load data
    df = pd.read_csv(data_path)
    df_clean = df.dropna()

    # Features
    features = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']
    X = df_clean[features].values

    # True labels
    y_true = df_clean['species']

    # Scaling
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # K-Means
    kmeans = KMeans(n_clusters=n_clusters, random_state=random_state, n_init=10)
    labels = kmeans.fit_predict(X_scaled)

    # Metrics
    silhouette = silhouette_score(X_scaled, labels)
    davies_bouldin = davies_bouldin_score(X_scaled, labels)
    calinski = calinski_harabasz_score(X_scaled, labels)

    ari = adjusted_rand_score(y_true, labels)
    nmi = normalized_mutual_info_score(y_true, labels)
    homogeneity = homogeneity_score(y_true, labels)

    inertia = kmeans.inertia_

    # Add cluster labels to dataframe
    df_clean['cluster'] = labels

    # Metrics DataFrame
    metrics_df = pd.DataFrame({
        "Metric": [
            "Silhouette Score",
            "Davies-Bouldin Score",
            "Calinski-Harabasz Score",
            "Adjusted Rand Index",
            "Normalized Mutual Information",
            "Homogeneity Score",
            "Inertia"
        ],
        "Value": [
            silhouette,
            davies_bouldin,
            calinski,
            ari,
            nmi,
            homogeneity,
            inertia
        ]
    })

    results = {
        'dataframe': df_clean,
        'model': kmeans,
        'scaler': scaler,
        'cluster_centers': scaler.inverse_transform(kmeans.cluster_centers_),
        'metrics_table': metrics_df
    }

    return results
````

````python
### Extract the Results
results = perform_kmeans_clustering("penguins.csv")
results['metrics_table']
````

The performance of the clustering model was evaluated using several **internal and external clustering evaluation metrics** to assess the quality, separation, and agreement of the clusters with the true species labels. The **Silhouette Score** obtained was **0.446**, which falls within the theoretical range of **–1 to 1**, where higher values indicate better defined clusters. A value of 0.446 suggests **moderate cluster separation**, meaning that most data points are reasonably well matched to their assigned clusters, although some overlap between clusters may still exist.

The **Davies–Bouldin Score** was **0.942**. This metric evaluates the similarity between clusters by comparing within-cluster dispersion with between-cluster separation. Since **lower values indicate better clustering**, a value below 1 suggests that the clusters are **relatively compact and well separated**.

The **Calinski–Harabasz Score** was **427.77**, which measures the ratio of between-cluster variance to within-cluster variance. Higher values indicate better clustering structure. The relatively high value obtained indicates that the clusters formed by the algorithm exhibit **good separation and internal cohesion**.

External validation metrics were also computed to compare the clustering results with the true species labels. The **Adjusted Rand Index (ARI)** of **0.799** indicates a **strong agreement** between the predicted clusters and the actual species classifications. Similarly, the **Normalized Mutual Information (NMI)** score of **0.790** shows a **high level of shared information** between the clustering assignments and the true labels.

Furthermore, the **Homogeneity Score** of **0.801** suggests that most clusters contain data points belonging primarily to a single species, indicating **high cluster purity**. Finally, the **Inertia value** of **370.77** represents the within-cluster sum of squared distances, reflecting relatively **compact clusters around their centroids**. Overall, these results demonstrate that the clustering model performs effectively in capturing the underlying structure of the dataset.


## 📊 **Silhouette Analysis for Clustering**

**Silhouette analysis** is a widely used method to **measure how well clusters are separated** and how **similar each point is to its own cluster compared to other clusters**. It is both **numeric** and **visual**, making it very useful for **cluster validation**.

---

# 🔍 **What is the Silhouette Score?**

For a point (i):

$$
s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}
$$

Where:

* (a(i)) = average distance of (i) to **all other points in the same cluster**
* (b(i)) = lowest average distance of (i) to points in **other clusters**

**Interpretation**:

| Silhouette Value | Meaning                               |
| ---------------- | ------------------------------------- |
| 1.0              | Perfectly clustered                   |
| 0.5–0.7          | Reasonable structure                  |
| ≈0               | On the border between clusters        |
| <0               | Misclassified or in the wrong cluster |

---

# ⚙️ **Silhouette Plot Concept**

* Each cluster is represented as a **horizontal bar**.
* Length of the bar = **silhouette value of each point**.
* Color = cluster label.
* A **longer bar** → better fit in the cluster.
* **Average silhouette score** gives a **global measure of clustering quality**.

---

# 💻 **Python Example (K-Means)**

```python
from sklearn.metrics import silhouette_samples, silhouette_score
```

## 📘 **Hierarchical Clustering**

### 🔍 **What is Hierarchical Clustering?**

Hierarchical clustering is an **unsupervised machine learning technique** used to group similar observations into clusters based on their similarity or distance.

Unlike methods such as **K-Means**, hierarchical clustering **does not require you to specify the number of clusters in advance**. Instead, it builds a **tree-like structure called a dendrogram**, which shows how observations are progressively merged or split into clusters.

The main idea is simple:
➡️ **Objects that are more similar are grouped together first**
➡️ **Objects that are less similar are merged later**

This approach is widely used in areas such as **bioinformatics, market segmentation, document classification, and gene expression analysis**.

---

# 🌳 **Types of Hierarchical Clustering**

Hierarchical clustering can be performed in two main ways:

## 1️⃣ **Agglomerative Clustering (Bottom-Up Approach)**

📌 This is the **most commonly used method**.

### 🔹 How it works

1. Start with **each observation as its own cluster**.
2. Compute the **distance between all clusters**.
3. Merge the **two closest clusters**.
4. Recompute distances between the new cluster and the remaining clusters.
5. Repeat until **all observations form a single cluster**.

### 📊 Step Illustration

| Step   | Cluster Structure                  |
| ------ | ---------------------------------- |
| Step 1 | Each data point = separate cluster |
| Step 2 | Closest points merge               |
| Step 3 | Small clusters merge               |
| Step 4 | Larger clusters form               |
| Step 5 | One final cluster                  |

📌 This process produces a **hierarchical tree structure**.

---

## 2️⃣ **Divisive Clustering (Top-Down Approach)**

📌 This is **less commonly used but conceptually opposite** to agglomerative clustering.

### 🔹 How it works

1. Start with **all observations in one cluster**.
2. Split the cluster into **two smaller clusters**.
3. Continue splitting clusters recursively.
4. Stop when each observation becomes its **own cluster**.

### 🧩 Simple Concept

```
All Data
   ↓
Two clusters
   ↓
Four clusters
   ↓
More refined clusters
```

---

# 📊 **Dendrogram (Cluster Tree)**

### 🌳 What is a Dendrogram?

A **dendrogram** is a **tree diagram used to visualize hierarchical clustering**.

It shows:

✔️ How clusters merge
✔️ At what **distance/similarity level** they merge
✔️ The **hierarchical relationships** between clusters

### 📌 How to Read a Dendrogram

🔹 **Leaves (bottom)** → Individual observations
🔹 **Branches** → Cluster merges
🔹 **Height of merge** → Distance between clusters

👉 **Lower merges = more similar observations**
👉 **Higher merges = less similar observations**

Example structure:

```
        ───── Cluster
       |
   ────|      ── Data point 1
  |    |
  |    └──── Data point 2
  |
  |         ── Data point 3
  └─────────|
            └─ Data point 4
```

---

# 📏 **Distance Measures Used**

Hierarchical clustering relies on **distance metrics** to determine similarity.

### 📐 Common Distance Measures

| Distance                  | Description                           |
| ------------------------- | ------------------------------------- |
| 📏 **Euclidean Distance** | Straight-line distance between points |
| 📊 **Manhattan Distance** | Sum of absolute differences           |
| 📉 **Minkowski Distance** | Generalized distance metric           |
| 🧬 **Cosine Similarity**  | Measures angle between vectors        |

Example (Euclidean distance):

$$
d(x,y)=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
$$

---

# 🔗 **Linkage Methods**

Linkage determines **how the distance between clusters is calculated**.

### Common Linkage Methods

| Linkage Method          | Description                       |
| ----------------------- | --------------------------------- |
| 🔗 **Single Linkage**   | Minimum distance between clusters |
| 📦 **Complete Linkage** | Maximum distance between clusters |
| 📊 **Average Linkage**  | Average distance between clusters |
| 🎯 **Ward’s Method**    | Minimizes within-cluster variance |

### Example Concept

Clusters A and B:

* **Single linkage:** closest points
* **Complete linkage:** farthest points
* **Average linkage:** average distance

---

# 🧠 **Advantages of Hierarchical Clustering**

✔️ No need to specify number of clusters in advance
✔️ Produces **interpretable dendrograms**
✔️ Works well with **small datasets**
✔️ Captures **nested cluster structures**

---

# ⚠️ **Limitations**

❌ Computationally expensive for **large datasets**
❌ Once clusters merge, they **cannot be undone**
❌ Sensitive to **noise and outliers**

## 📦 **DBSCAN (Density-Based Spatial Clustering of Applications with Noise)**

### 🔍 **What is DBSCAN?**

**DBSCAN** is an **unsupervised clustering algorithm** that groups together data points that lie in **dense regions** of a dataset.

Unlike algorithms such as **K-Means**, DBSCAN does **not require the number of clusters to be specified beforehand**. Instead, it identifies clusters based on **density**, meaning how closely packed the data points are.

📌 In simple terms:

* 🟢 **Dense regions → Form clusters**
* 🔴 **Sparse regions → Considered noise (outliers)**

This makes DBSCAN very powerful when dealing with **irregular cluster shapes and noisy data**.

---

# 🧠 **Key Idea Behind DBSCAN**

DBSCAN works by identifying areas where **many data points are close together**.

The algorithm expands clusters by connecting **points that are density-reachable** from each other.

👉 Points that are **too far from dense areas are labeled as noise**.

---

# ⚙️ **Two Important Parameters**

DBSCAN relies on **two main parameters**.

### 📏 1️⃣ **Epsilon (ε)** – Neighborhood Radius

This defines the **maximum distance** between two points for them to be considered neighbors.

📌 Interpretation:

* 🔹 Small ε → Many small clusters
* 🔹 Large ε → Fewer, larger clusters

---

### 🔢 2️⃣ **MinPts (Minimum Points)**

This is the **minimum number of points required to form a dense region**.

📌 Interpretation:

* If a point has **at least MinPts neighbors within ε**, it becomes a **core point**.

Typical rule:

$$
MinPts \geq \text{Dimension} + 1
$$

---

# 📍 **Types of Points in DBSCAN**

DBSCAN classifies data points into **three categories**.

### 🟢 **1. Core Points**

A point is a **core point** if it has **at least MinPts neighbors within ε distance**.

✔ Located in **dense regions**
✔ Responsible for forming clusters

---

### 🟡 **2. Border Points**

A **border point** lies within the neighborhood of a **core point** but does **not have enough neighbors** to be a core point itself.

✔ Belongs to a cluster
❌ Cannot expand the cluster

---

### 🔴 **3. Noise Points (Outliers)**

Noise points are points that **do not belong to any cluster**.

✔ Too far from dense regions
✔ Treated as **outliers**

---

# 🔄 **How DBSCAN Works (Step-by-Step)**

### 🧩 Algorithm Process

1️⃣ Choose a random point from the dataset.

2️⃣ Find all points within distance **ε**.

3️⃣ If the number of neighbors ≥ **MinPts**, a **cluster is formed**.

4️⃣ Expand the cluster by including all **density-reachable points**.

5️⃣ Continue until no more points can be added.

6️⃣ Repeat the process for remaining points.

The **Silhouette Score** measures how well each data point fits within its assigned cluster compared to other clusters.

It evaluates two things:

* 📏 **Cohesion** – how close points are within the same cluster
* 📐 **Separation** – how far clusters are from each other

The score ranges between:

$$
-1 \leq S \leq 1
$$

| ----------------- | -------------------- |

# 🔁 **2. Adjusted Rand Index (ARI)**

### 🔍 What is ARI?

The **Adjusted Rand Index** measures the similarity between:

* 📊 **True class labels**
* 📦 **Predicted cluster assignments**

It evaluates how well the clustering algorithm matches the **actual grouping of the data**.

The ARI adjusts for **random clustering agreement**, making it more reliable.

### 📊 Range of ARI

$$
-1 \leq ARI \leq 1
$$

| ARI Value        | Interpretation     |
| ---------------- | ------------------ |
| 🟢 **1**         | Perfect clustering |
| 🟡 **0.5 – 0.8** | Moderate agreement |
| 🟠 **0 – 0.5**   | Weak agreement     |
| 🔴 **< 0**       | Worse than random  |

# 📌 **DIMENSION REDUCTION**

Dimension reduction is a **technique to reduce the number of input variables (features) in a dataset** while retaining most of the relevant information. It is widely used in **machine learning, data visualization, and noise reduction**.

---

### 🛠 **Why Dimension Reduction?**

1. **Curse of Dimensionality**

   * High-dimensional data can lead to sparse observations, making models less effective.
2. **Improved Model Performance**

   * Reduces overfitting and computational cost.
3. **Visualization**

   * Projects high-dimensional data into **2D or 3D** for plotting and analysis.
4. **Noise Reduction**

   * Removes irrelevant or redundant features.

---

### ⚙️ **Types of Dimension Reduction**

#### 1️⃣ Feature Selection

* **Definition:** Selects a subset of original features based on importance.
* **Methods:**

  * Filter methods (e.g., correlation threshold, chi-square test)
  * Wrapper methods (e.g., forward/backward selection)
  * Embedded methods (e.g., Lasso regression)
* **Key Idea:** Keeps original features; no new features are created.

#### 2️⃣ Feature Extraction

* **Definition:** Transforms data into **new feature space** (combines original features).
* **Methods:**

  * **Principal Component Analysis (PCA)** – Projects data onto orthogonal axes with maximum variance.
  * **Linear Discriminant Analysis (LDA)** – Maximizes class separability.
  * **t-SNE** and **UMAP** – Non-linear techniques for visualization in 2D/3D.
* **Key Idea:** Creates **new features** summarizing original ones.

---

### 📊 **Principal Component Analysis (PCA)**

* Projects data onto a set of **orthogonal axes** (principal components).
* **Steps:**

  1. Standardize features.
  2. Compute covariance matrix.
  3. Calculate eigenvectors and eigenvalues.
  4. Select top **k components** covering most variance.
  5. Transform original data onto these components.
* **Applications:** Noise reduction, clustering, visualization.

---

### 🔍 **Linear Discriminant Analysis (LDA)**

* Supervised method; maximizes **between-class variance** and minimizes **within-class variance**.
* Often used in **classification tasks**.

---

### 🎯 **t-SNE & UMAP**

* **t-SNE (t-Distributed Stochastic Neighbor Embedding):**

  * Non-linear; preserves local structure.
  * Good for **visualizing clusters in 2D/3D**.
* **UMAP (Uniform Manifold Approximation and Projection):**

  * Faster than t-SNE; preserves both **local and global structure**.

---

### ✅ **Key Points**

* Reduces **computational cost** and **overfitting**.
* Helps in **visualization** of high-dimensional data.
* Must balance **information retention vs. dimensionality reduction**.
* Choice of method depends on **data type, supervised/unsupervised task, and interpretability**.

# 📘 **t-SNE (t-Distributed Stochastic Neighbor Embedding)**

## 🔍 **What is t-SNE?**

**t-SNE** is a **non-linear dimensionality reduction technique** used mainly for **visualizing high-dimensional data** in **two or three dimensions**.

In many datasets (especially in **machine learning, bioinformatics, and image processing**), data may contain **many features or variables**. Visualizing such data directly is difficult. t-SNE solves this problem by **transforming high-dimensional data into a lower-dimensional space while preserving meaningful patterns**.

📌 In simple terms:

* 📊 **High-dimensional data → Reduced to 2D or 3D**
* 🔎 **Similar points stay close together**
* 🧩 **Hidden clusters become visible**

---

# 🧠 **Main Idea Behind t-SNE**

t-SNE focuses on preserving the **local structure of the data**.

This means:

✔ Points that are **similar in high dimensions remain close together**
✔ Points that are **very different remain far apart**

The algorithm converts **similarities between data points into probabilities**, then tries to reproduce these probabilities in a lower-dimensional space.

---

# ⚙️ **How t-SNE Works (Conceptually)**

### Step-by-Step Process

1️⃣ 📊 **Compute pairwise similarities** between data points in the high-dimensional space.

2️⃣ 🎯 Convert these similarities into **probability distributions**.

3️⃣ 📉 Map the data into **lower dimensions (2D or 3D)**.

4️⃣ 🔄 Adjust the positions of points iteratively to **minimize the difference between the two probability distributions**.

5️⃣ 📍 Points with similar neighbors stay **close together** in the final visualization.

---

# 🔑 **Key Features of t-SNE**

### 🧩 1. Non-Linear Dimensionality Reduction

Unlike **linear methods such as PCA**, t-SNE can capture **complex non-linear relationships** in data.

✔ Detects **curved or irregular structures**
✔ Useful for **complex datasets**

---

### 🔎 2. Preserves Local Structure

t-SNE focuses on **local neighborhoods**.

✔ Nearby points remain close
✔ Clusters become visible

This makes it excellent for **cluster visualization**.

---

### 📊 3. Visualization Tool

t-SNE is mainly used for **exploratory data analysis and visualization**, not for predictive modeling.

Common visualizations include:

* 📈 **2D scatter plots**
* 🌐 **3D cluster plots**

---

# 📌 **Important Parameter**

### 🎛 **Perplexity**

Perplexity controls how many **nearest neighbors** are considered during clustering.

Typical values:

| Value    | Interpretation                  |
| -------- | ------------------------------- |
| 🔹 5–10  | Focus on very local structure   |
| 🔹 20–50 | Balanced local/global structure |
| 🔹 >50   | Captures broader structure      |

Common default:

👉 **Perplexity = 30**

## 📊 **Cluster Stability Analysis**

Cluster stability analysis evaluates how **consistent clustering results remain when the dataset is slightly changed**, such as through **bootstrap resampling**. A stable clustering algorithm should produce **similar cluster assignments even when the data varies slightly**.

In your case, stability was assessed using **bootstrap sampling and the Adjusted Rand Index (ARI)**.

---

# 🔍 **Method Used**

### 🧩 Bootstrap Stability Procedure

The stability analysis follows these steps:

1️⃣ 🔄 **Bootstrap Sampling**
A new dataset is created by randomly sampling observations from the original dataset **with replacement**.

2️⃣ 📦 **Clustering on Original Data**
The clustering algorithm (here **K-Means with 3 clusters**) is applied to the original dataset.

3️⃣ 📦 **Clustering on Bootstrap Sample**
The same clustering algorithm is applied to the resampled dataset.

4️⃣ 🔁 **Cluster Comparison**
The **Adjusted Rand Index (ARI)** is used to compare the cluster assignments from the two datasets.

5️⃣ 📊 **Repeat Multiple Times**
This process is repeated **several times (10 bootstrap iterations)**.

6️⃣ 📈 **Compute Stability Statistics**
The **mean and standard deviation of ARI scores** are calculated.

ARI measures similarity between two cluster assignments:

[

## 📊 **Cluster Stability Heatmap**

A **stability heatmap** is used to visualize how **consistently observations are grouped together across multiple bootstrap clustering runs**.

Instead of looking at a single stability score, this method shows **pairwise stability**:

* 🟢 **High value (≈1)** → Two points are **almost always clustered together**
* 🟡 **Medium value (~0.5)** → Sometimes grouped together
* 🔴 **Low value (≈0)** → Rarely placed in the same cluster

This produces a **co-clustering matrix**, which can be visualized using a **heatmap**.

---

# 🧠 **Concept of Stability Heatmap**

The idea is to compute a **consensus matrix**:

$$
C_{ij} = \frac{\text{Number of times } i \text{ and } j \text{ appear in the same cluster}}{\text{Total bootstrap runs}}
$$

Where:

* (C_{ij}) measures **cluster stability between point i and j**.

---

# 💻 **Python Code for Stability Heatmap**

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.utils import resample

def stability_heatmap(X, n_clusters=3, n_bootstrap=50):

    n = X.shape[0]
    consensus = np.zeros((n, n))

    for _ in range(n_bootstrap):

        # Bootstrap sample indices
        indices = resample(range(n), n_samples=n)

        X_sample = X[indices]

        # Cluster bootstrap sample
        kmeans = KMeans(n_clusters=n_clusters, n_init=10)
        labels = kmeans.fit_predict(X_sample)

        # Update consensus matrix
        for i in range(n):
            for j in range(n):
                if labels[i] == labels[j]:
                    consensus[indices[i], indices[j]] += 1

    consensus = consensus / n_bootstrap

    return consensus


# Compute stability matrix

# Plot heatmap

plt.title("Cluster Stability Heatmap (Consensus Matrix)")

plt.ylabel("Samples")
plt.show()
```

### **✅ 4. Cluster Tendency (Is Clustering Even Valid?)**
Before clustering, ask:

❓ "Does my data have clusters at all?"

🔹 Hopkins Test (Measures clustering tendency)

````python
def hopkins_statistic(X, sample_size=10):
    d = X.shape[1]
    n = len(X)
    sample_size = min(sample_size, n)

    # Random points in the space
    rand_sample = np.random.uniform(X.min(axis=0), X.max(axis=0), (sample_size, d))

    # Distances from random points to nearest real point
    from sklearn.neighbors import NearestNeighbors
    nbrs = NearestNeighbors(n_neighbors=1).fit(X)
    _, dist_real = nbrs.kneighbors(rand_sample)

    # Distances from real points to nearest other real point
    _, dist_self = nbrs.kneighbors(X[:sample_size])

    # Hopkins statistic
    H = dist_real.sum() / (dist_real.sum() + dist_self.sum())
    return H.item()

H = hopkins_statistic(X_scaled)
print(f"Hopkins Statistic: {H:.3f}")
print("→ Data is " + ("clustered" if H > 0.7 else "not clustered"))
````

## **Cluster Ensemble Model**

A **Cluster Ensemble Model** (also called **Consensus Clustering**) combines several clustering algorithms to produce a **more stable and robust final clustering**.

Instead of relying on one algorithm, we combine results from:

* **K-Means**
* **Agglomerative Hierarchical Clustering**
* **DBSCAN**

This approach is widely used in **bioinformatics, image analysis, and machine learning research**.

---

# 🧠 Concept of Cluster Ensemble

Each algorithm produces cluster labels:

| Sample | K-Means | Hierarchical | DBSCAN |
| ------ | ------- | ------------ | ------ |
| 1      | 0       | 1            | 0      |
| 2      | 2       | 2            | 1      |
| 3      | 1       | 0            | -1     |

We combine these labels to produce a **consensus cluster**.

Two common approaches:

1️⃣ **Majority Voting**
2️⃣ **Meta-Clustering (classifier on cluster outputs)**

We will implement the **meta-clustering approach**.

---

# 📊 Step 1: Fit Multiple Clustering Models

```python
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN

# K-Means
kmeans = KMeans(n_clusters=3, random_state=123)
kmeans_labels = kmeans.fit_predict(X_scaled)

# Hierarchical
hierarchical = AgglomerativeClustering(n_clusters=3)
hier_labels = hierarchical.fit_predict(X_scaled)

# DBSCAN
dbscan = DBSCAN(eps=0.7, min_samples=5)
db_labels = dbscan.fit_predict(X_scaled)
```

---

# 📊 Step 2: Combine Cluster Labels

Create a dataset of cluster assignments.

```python
import numpy as np

cluster_matrix = np.column_stack([
    kmeans_labels,
    hier_labels,
    db_labels
])
```

Example structure:

| KMeans | Hierarchical | DBSCAN |
| ------ | ------------ | ------ |
| 0      | 1            | 0      |
| 2      | 2            | 1      |
| 1      | 0            | -1     |

---

# 📊 Step 3: Train Consensus Model

Use a classifier to learn **combined cluster structure**.

```python
from sklearn.ensemble import RandomForestClassifier

ensemble_model = RandomForestClassifier(random_state=123)
ensemble_model.fit(cluster_matrix, kmeans_labels)
```

Here:

* cluster outputs → predictors
* final cluster → target
